```python

import gym
import gym_super_mario_bros
import cv2
import numpy as np
import matplotlib.pyplot as plt

from nes_py.wrappers import JoypadSpace
from gym_super_mario_bros.actions import SIMPLE_MOVEMENT

# Recreate the Mario environment
env = JoypadSpace(
    gym_super_mario_bros.make("SuperMarioBros-v0"),
    SIMPLE_MOVEMENT
)

# Capture one frame
obs = env.reset()

# Preprocess the image
gray = cv2.cvtColor(obs, cv2.COLOR_RGB2GRAY)
small_frame = cv2.resize(gray, (84, 84))
processed_frame = small_frame.astype(np.float32) / 255.0

print("Original shape:", obs.shape)
print("Processed shape:", processed_frame.shape)

plt.imshow(processed_frame, cmap="gray")
plt.axis("off")
plt.show()
```

```python

import torch
import torch.nn as nn
import torch.nn.functional as F

class MarioCNN(nn.Module):
    def __init__(self, num_actions=7):
        super().__init__()

        self.conv1 = nn.Conv2d(1, 32, kernel_size=8, stride=4)
        self.conv2 = nn.Conv2d(32, 64, kernel_size=4, stride=2)
        self.conv3 = nn.Conv2d(64, 64, kernel_size=3, stride=1)

        self.fc1 = nn.Linear(64 * 7 * 7, 512)
        self.fc2 = nn.Linear(512, num_actions)

    def forward(self, x):
        x = F.relu(self.conv1(x))
        x = F.relu(self.conv2(x))
        x = F.relu(self.conv3(x))

        x = x.reshape(x.shape[0], -1)
        x = F.relu(self.fc1(x))
        return self.fc2(x)


device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

model = MarioCNN(num_actions=7).to(device)

print("Device:", device)
print("Parameters:", sum(p.numel() for p in model.parameters()))
```

```python

from collections import deque
import random
import numpy as np

class ReplayBuffer:
    def __init__(self, capacity=10_000):
        self.memory = deque(maxlen=capacity)

    def add(self, state, action, reward, next_state, done):
        self.memory.append(
            (state, action, reward, next_state, done)
        )

    def sample(self, batch_size):
        batch = random.sample(self.memory, batch_size)

        states, actions, rewards, next_states, dones = zip(*batch)

        return (
            np.array(states, dtype=np.float32),
            np.array(actions, dtype=np.int64),
            np.array(rewards, dtype=np.float32),
            np.array(next_states, dtype=np.float32),
            np.array(dones, dtype=np.float32)
        )

    def __len__(self):
        return len(self.memory)


replay_buffer = ReplayBuffer()

print("Replay buffer created!")
print("Current experiences:", len(replay_buffer))
```

```python

import cv2
import numpy as np

def preprocess(obs):
    gray = cv2.cvtColor(obs, cv2.COLOR_RGB2GRAY)
    resized = cv2.resize(gray, (84, 84))
    return resized.astype(np.float32) / 255.0

obs = env.reset()
state = preprocess(obs)

total_reward = 0

for step in range(200):
    action = env.action_space.sample()

    next_obs, reward, done, info = env.step(action)
    next_state = preprocess(next_obs)

    replay_buffer.add(
        state, action, reward, next_state, done
    )

    state = next_state
    total_reward += reward

    if done:
        obs = env.reset()
        state = preprocess(obs)

print("Experiences collected:", len(replay_buffer))
print("Total reward:", total_reward)
```

```python

import torch
import torch.nn as nn
import torch.optim as optim
import numpy as np

# 1. Create a target network
target_model = MarioCNN(num_actions=7).to(device)
target_model.load_state_dict(model.state_dict())
target_model.eval()

# 2. Configure learning
optimizer = optim.Adam(model.parameters(), lr=1e-4)
loss_fn = nn.SmoothL1Loss()

gamma = 0.99
batch_size = 32

# 3. Sample past experiences
states, actions, rewards, next_states, dones = (
    replay_buffer.sample(batch_size)
)

states = torch.tensor(states, dtype=torch.float32).unsqueeze(1).to(device)
actions = torch.tensor(actions, dtype=torch.long).to(device)
rewards = torch.tensor(rewards, dtype=torch.float32).to(device)
next_states = torch.tensor(next_states, dtype=torch.float32).unsqueeze(1).to(device)
dones = torch.tensor(dones, dtype=torch.float32).to(device)

# 4. Predict Q-values for the actions actually taken
model.train()
current_q = model(states).gather(1, actions.unsqueeze(1)).squeeze(1)

# 5. Calculate target Q-values
with torch.no_grad():
    next_q = target_model(next_states).max(dim=1)[0]
    target_q = rewards + gamma * next_q * (1 - dones)

# 6. Update network using backpropagation
loss = loss_fn(current_q, target_q)

optimizer.zero_grad()
loss.backward()
torch.nn.utils.clip_grad_norm_(model.parameters(), 10.0)
optimizer.step()

print("First DQN update completed!")
print("Batch size:", batch_size)
print("Loss:", round(loss.item(), 4))
```

```python

import random
import numpy as np
import cv2
import torch
import torch.nn.functional as F
import matplotlib.pyplot as plt

# Create the environment if it doesn't exist
try:
    env
except NameError:
    import gym_super_mario_bros
    from nes_py.wrappers import JoypadSpace
    from gym_super_mario_bros.actions import SIMPLE_MOVEMENT

    env = JoypadSpace(
        gym_super_mario_bros.make("SuperMarioBros-v0"),
        SIMPLE_MOVEMENT
    )

# Image preprocessing
def preprocess(obs):
    gray = cv2.cvtColor(obs, cv2.COLOR_RGB2GRAY)
    gray = cv2.resize(gray, (84, 84))
    return gray.astype(np.float32) / 255.0

# Keep training lightweight initially
num_steps = 500
batch_size = 32
epsilon = 1.0
epsilon_min = 0.1
epsilon_decay = 0.995
gamma = 0.99

obs = env.reset()
state = preprocess(obs)

episode_reward = 0.0
episode_rewards = []
losses = []

model.train()

for step in range(1, num_steps + 1):

    # Explore randomly or use the CNN
    if random.random() < epsilon:
        action = env.action_space.sample()
    else:
        state_tensor = torch.as_tensor(
            state, dtype=torch.float32, device=device
        )[None, None, :, :]

        with torch.no_grad():
            action = model(state_tensor).argmax(dim=1).item()

    next_obs, reward, done, info = env.step(action)
    next_state = preprocess(next_obs)

    replay_buffer.add(
        state, action, reward, next_state, done
    )

    state = next_state
    episode_reward += float(reward)

    # Learn from replayed experiences
    if len(replay_buffer) >= batch_size and step % 4 == 0:
        states, actions, rewards, next_states, dones = (
            replay_buffer.sample(batch_size)
        )

        states = torch.as_tensor(
            states, dtype=torch.float32, device=device
        ).unsqueeze(1)

        next_states = torch.as_tensor(
            next_states, dtype=torch.float32, device=device
        ).unsqueeze(1)

        actions = torch.as_tensor(
            actions, dtype=torch.long, device=device
        )
        rewards = torch.as_tensor(
            rewards, dtype=torch.float32, device=device
        )
        dones = torch.as_tensor(
            dones, dtype=torch.float32, device=device
        )

        predicted = model(states).gather(
            1, actions.unsqueeze(1)
        ).squeeze(1)

        with torch.no_grad():
            future = target_model(next_states).max(dim=1)[0]
            target = rewards + gamma * future * (1 - dones)

        loss = F.smooth_l1_loss(predicted, target)

        optimizer.zero_grad()
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 10.0)
        optimizer.step()
        losses.append(loss.item())

    if step % 100 == 0:
        target_model.load_state_dict(model.state_dict())

    epsilon = max(epsilon_min, epsilon * epsilon_decay)

    if done:
        episode_rewards.append(episode_reward)
        print(
            f"Step {step} | Episode reward: {episode_reward:.1f} "
            f"| Epsilon: {epsilon:.2f}"
        )

        obs = env.reset()
        state = preprocess(obs)
        episode_reward = 0.0

    if step % 50 == 0:
        print(
            f"Progress: {step}/{num_steps} | "
            f"Memory: {len(replay_buffer)} | Epsilon: {epsilon:.2f}"
        )

# Save the current model checkpoint
torch.save({
    "model": model.state_dict(),
    "target_model": target_model.state_dict(),
    "optimizer": optimizer.state_dict(),
    "epsilon": epsilon,
}, "/content/mario_dqn_checkpoint.pt")

print("Training session finished.")
print("Checkpoint saved to /content/mario_dqn_checkpoint.pt")
print("Learning updates:", len(losses))

if losses:
    print("Final loss:", round(losses[-1], 4))

if episode_rewards:
    plt.plot(episode_rewards, marker="o")
    plt.xlabel("Episode")
    plt.ylabel("Total reward")
    plt.title("Mario Training Rewards")
    plt.show()
else:
    print("No complete episodes yet; continue training to evaluate episodes.")
```

```python

import torch
import torch.nn as nn
import torch.nn.functional as F

class MarioCNN(nn.Module):
    def __init__(self, num_actions=7):
        super().__init__()

        # Four consecutive grayscale frames
        self.conv1 = nn.Conv2d(4, 32, kernel_size=8, stride=4)
        self.conv2 = nn.Conv2d(32, 64, kernel_size=4, stride=2)
        self.conv3 = nn.Conv2d(64, 64, kernel_size=3, stride=1)

        self.fc1 = nn.Linear(64 * 7 * 7, 512)
        self.fc2 = nn.Linear(512, num_actions)

    def forward(self, x):
        x = F.relu(self.conv1(x))
        x = F.relu(self.conv2(x))
        x = F.relu(self.conv3(x))
        x = x.reshape(x.shape[0], -1)
        x = F.relu(self.fc1(x))
        return self.fc2(x)

model = MarioCNN(num_actions=7).to(device)
target_model = MarioCNN(num_actions=7).to(device)
target_model.load_state_dict(model.state_dict())

optimizer = torch.optim.Adam(model.parameters(), lr=1e-4)

print("Four-frame Mario CNN ready!")
print("Input shape required: [batch, 4, 84, 84]")
print("Output shape: [batch, 7]")
```

```python

from collections import deque
import random
import numpy as np
import cv2

def preprocess(obs):
    gray = cv2.cvtColor(obs, cv2.COLOR_RGB2GRAY)
    gray = cv2.resize(gray, (84, 84))
    return gray.astype(np.uint8)

class FrameStack:
    def __init__(self, num_frames=4):
        self.frames = deque(maxlen=num_frames)

    def reset(self, frame):
        self.frames.clear()
        for _ in range(4):
            self.frames.append(frame)
        return np.stack(self.frames, axis=0)

    def add(self, frame):
        self.frames.append(frame)
        return np.stack(self.frames, axis=0)

class ReplayBuffer:
    def __init__(self, capacity=3000):
        self.memory = deque(maxlen=capacity)

    def add(self, state, action, reward, next_state, done):
        self.memory.append(
            (state, action, reward, next_state, done)
        )

    def sample(self, batch_size):
        batch = random.sample(self.memory, batch_size)
        states, actions, rewards, next_states, dones = zip(*batch)

        return (
            np.stack(states).astype(np.uint8),
            np.array(actions, dtype=np.int64),
            np.array(rewards, dtype=np.float32),
            np.stack(next_states).astype(np.uint8),
            np.array(dones, dtype=np.float32)
        )

    def __len__(self):
        return len(self.memory)

replay_buffer = ReplayBuffer()

print("Four-frame replay buffer ready!")
print("Capacity:", replay_buffer.memory.maxlen)
```

```python

import torch
import torch.nn.functional as F
import numpy as np

batch_size = 32
gamma = 0.99
training_updates = 200
losses = []

model.train()

for update in range(training_updates):
    states, actions, rewards, next_states, dones = (
        replay_buffer.sample(batch_size)
    )

    # Normalize images to [0, 1]
    states = torch.as_tensor(
        states, dtype=torch.float32, device=device
    ) / 255.0

    next_states = torch.as_tensor(
        next_states, dtype=torch.float32, device=device
    ) / 255.0

    actions = torch.as_tensor(
        actions, dtype=torch.long, device=device
    )
    rewards = torch.as_tensor(
        rewards, dtype=torch.float32, device=device
    )
    dones = torch.as_tensor(
        dones, dtype=torch.float32, device=device
    )

    # Q-value for the action taken
    predicted_q = model(states).gather(
        1, actions.unsqueeze(1)
    ).squeeze(1)

    # Target Q-value
    with torch.no_grad():
        future_q = target_model(next_states).max(dim=1)[0]
        target_q = rewards + gamma * future_q * (1 - dones)

    # Backpropagation
    loss = F.smooth_l1_loss(predicted_q, target_q)

    optimizer.zero_grad()
    loss.backward()
    torch.nn.utils.clip_grad_norm_(model.parameters(), 10.0)
    optimizer.step()

    losses.append(loss.item())

    # Periodically synchronize target network
    if (update + 1) % 25 == 0:
        target_model.load_state_dict(model.state_dict())

    if (update + 1) % 50 == 0:
        print(
            f"Update {update + 1}/{training_updates} | "
            f"Loss: {loss.item():.4f}"
        )

print("Four-frame training completed!")
print("Average loss:", round(float(np.mean(losses)), 4))
```
